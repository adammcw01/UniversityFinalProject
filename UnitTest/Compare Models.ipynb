This notebook loads and tests each of the models in a given directory

```python
from glob import glob
import tensorflow as tf
import numpy as np
import os
```

```python
path = r"models"
models = glob(os.path.join(path, "*"))
models.sort()
```

```python
testData = r"E:\QUB\Stage3\CSC3002-Project\nsynth-data\nsynth-test\GeneratedDataset"
dataset = tf.data.Dataset.load(testData)
```

```python
def testModel(modelPath, dataset):
    model = tf.keras.models.load_model(modelPath)
    print(os.path.basename(modelPath))
    model.summary()
    print('\n')
    
    predicitons = model.predict(dataset.batch(128), verbose=0)
    _, labels = list(zip(*dataset))

    accuracyTest = [labels[i].numpy() == np.argmax(p) for i, p in enumerate(predicitons)]
    return round((accuracyTest.count(True)/len(accuracyTest))*100, 2)
```

```python
accuracies = [testModel(model, dataset) for model in models]
```
